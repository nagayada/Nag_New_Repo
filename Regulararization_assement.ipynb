### Part 1: Breast Cancer Classification (Questions 1, 2, and 3)

In this section, we will load the breast cancer dataset, split it (80:20), scale features using `MinMaxScaler`, fit a logistic regression model, and generate a classification report.

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report

# 1. Load the breast cancer dataset
cancer_data = load_breast_cancer()
X_cancer = cancer_data.data
y_cancer = cancer_data.target

# 2. Split dataset with an 80:20 ratio and random_state = 9001
X_train_c, X_test_c, y_train_c, y_test_c = train_test_split(
    X_cancer, y_cancer, test_size=0.2, random_state=9001
)

# 3. Scale the data using MinMaxScaler
scaler_c = MinMaxScaler()
X_train_c_scaled = scaler_c.fit_transform(X_train_c)
X_test_c_scaled = scaler_c.transform(X_test_c)

# 4. Fit Logistic Regression using default solver
log_reg = LogisticRegression()
log_reg.fit(X_train_c_scaled, y_train_c)

# 5. Predict and display classification report
y_pred_c = log_reg.predict(X_test_c_scaled)
print("Classification Report for Breast Cancer Dataset:")
print(classification_report(y_test_c, y_pred_c))
```

```python
import numpy as np

# Target features specified by the user
target_features = ['mean fractal dimension', 'area error', 'mean perimeter', 'worst radius']

# Map dataset feature names to their indices and coefficients
feature_names = list(cancer_data.feature_names)
coefs = log_reg.coef_[0]

# Filter for specified features and find their magnitude
specified_coefs = {}
for feat in target_features:
    if feat in feature_names:
        idx = feature_names.index(feat)
        specified_coefs[feat] = coefs[idx]

# Display the coefficients and identify the highest magnitude
print("Coefficients for the requested features:")
for feat, val in specified_coefs.items():
    print(f"  {feat}: {val:.4f} (magnitude: {abs(val):.4f})")

highest_feature = max(specified_coefs, key=lambda k: abs(specified_coefs[k]))
print(f"\nThe feature with the highest magnitude coefficient among them is: '{highest_feature}' with a value of {specified_coefs[highest_feature]:.4f}")
```

### Part 2: Diabetes Disease Progression Regression (Questions 4 and 5)

In this section, we will load the diabetes dataset, split it (80:20), scale features using `MinMaxScaler`, and compare three regression models: Linear Regression (no regularization), Lasso (L1 regularization), and Ridge (L2 regularization) with `alpha=1`.

```python
import numpy as np
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.metrics import mean_squared_error

# 1. Load the diabetes dataset
diabetes_data = load_diabetes()
X_diab = diabetes_data.data
y_diab = diabetes_data.target

# 2. Split dataset with an 80:20 ratio and random_state = 9001
X_train_d, X_test_d, y_train_d, y_test_d = train_test_split(
    X_diab, y_diab, test_size=0.2, random_state=9001
)

# 3. Scale the data using MinMaxScaler
scaler_d = MinMaxScaler()
X_train_d_scaled = scaler_d.fit_transform(X_train_d)
X_test_d_scaled = scaler_d.transform(X_test_d)

# 4. Initialize the three models
models = {
    "No Regularization (Linear Regression)": LinearRegression(),
    "L1 Regularization (Lasso, alpha=1)": Lasso(alpha=1.0),
    "L2 Regularization (Ridge, alpha=1)": Ridge(alpha=1.0)
}

# 5. Fit, predict, evaluate MSE and print coefficients for each model
for name, model in models.items():
    model.fit(X_train_d_scaled, y_train_d)
    preds = model.predict(X_test_d_scaled)
    mse = mean_squared_error(y_test_d, preds)
    print(f"=== {name} ===")
    print(f"Mean Squared Error (MSE): {mse:.4f}")
    print("Coefficients:")
    for feature, coef in zip(diabetes_data.feature_names, model.coef_):
        print(f"  {feature}: {coef:.4f}")
    print("\n")
```

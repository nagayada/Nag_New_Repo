```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_squared_error, mean_absolute_error

# Load the dataset
file_path = '/content/f4aa1250-3c0a-480b-bd6b-c953fa0e987c-College.xlsx'
df = pd.read_excel(file_path)

# Display initial info to understand structure
display(df.head())
print(df.info())
```

```python
# Remove the column containing the name of each university
# Assuming the first column is the name, we drop it
df_cleaned = df.iloc[:, 1:].copy()

# Obtain descriptive statistics on the numeric columns and round to 2 decimal places
descriptive_stats = df_cleaned.describe().round(2)
display(descriptive_stats)
```

```python
# One-hot encode the 'Private' variable, setting drop_first=True
df_encoded = pd.get_dummies(df_cleaned, columns=['Private'], drop_first=True)

# Define features (X) and target (y)
y = df_encoded['Grad.Rate']
X = df_encoded.drop(columns=['Grad.Rate'])

# Split the dataset into 70:30 train:test set with random_state=42
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.30, random_state=42)

# Fit the model on the training set
model = LinearRegression()
model.fit(X_train, y_train)

# Output model coefficients, rounded to 2 decimal places
coefficients = pd.Series(model.coef_, index=X.columns).round(2)
print("Model Coefficients:")
display(coefficients)
```

```python
# Obtain predictions on both training and testing sets
y_train_pred = model.predict(X_train)
y_test_pred = model.predict(X_test)

# Helper function to compute and round metrics
def get_metrics(y_true, y_pred):
    r2 = round(r2_score(y_true, y_pred), 2)
    mse = round(mean_squared_error(y_true, y_pred), 2)
    mae = round(mean_absolute_error(y_true, y_pred), 2)
    return r2, mse, mae

r2_train, mse_train, mae_train = get_metrics(y_train, y_train_pred)
r2_test, mse_test, mae_test = get_metrics(y_test, y_test_pred)

# Display metrics
metrics_df = pd.DataFrame({
    'Metric': ['R²', 'Mean Squared Error (MSE)', 'Mean Absolute Error (MAE)'],
    'Train Set': [r2_train, mse_train, mae_train],
    'Test Set': [r2_test, mse_test, mae_test]
})
display(metrics_df)
```

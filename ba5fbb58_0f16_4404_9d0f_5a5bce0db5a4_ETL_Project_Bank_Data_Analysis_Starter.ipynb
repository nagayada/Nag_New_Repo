````python
# Initialize Spark session

# URL of the Wikipedia page

# Use pandas to read all HTML tables from the page

# Load the correct table (Table 1)

# Convert pandas DataFrame to PySpark DataFrame

# Rename columns

# Show the first few rows of the PySpark DataFrame

# Print the schema of the DataFrame

# Configure logging
import logging
import pandas as pd
from pyspark.sql import SparkSession
from pyspark.sql.types import DoubleType

# Configure logging
logging.basicConfig(
    level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s"
)
logger = logging.getLogger(__name__)

# Initialize Spark session
spark = SparkSession.builder.appName("LargestBanksMarketCap").getOrCreate()
logger.info("Spark Session successfully initialized.")

# URL of the Wikipedia page
url = "https://web.archive.org/web/20230908091635/https://en.wikipedia.org/wiki/List_of_largest_banks"

try:
    tables = pd.read_html(url)
    logger.info(f"Successfully fetched {len(tables)} tables from the page.")
except Exception as e:
    logger.error(f"Failed to fetch data from URL: {e}")
    spark.stop()
    exit(1)

# Load the correct table (Table 1)
# The 'By market capitalization' table sits at index 0
pd_df = tables[0]

# Pre-cleaning: Ensure all columns are strings before sending to PySpark
pd_df.columns = [str(col) for col in pd_df.columns]
pd_df = pd_df.astype(str)

# Convert pandas DataFrame to PySpark DataFrame
spark_df = spark.createDataFrame(pd_df)
logger.info("Pandas DataFrame converted to PySpark DataFrame.")

# Rename columns
# Standardizing names for clean Spark SQL execution
spark_df = spark_df.withColumnRenamed("Rank", "rank") \
                   .withColumnRenamed("Bank name", "bank_name") \
                   .withColumnRenamed("Market cap (US$ billion)", "market_cap_usd_billion")

# Minor data type fix: Convert the market cap values from String to Double
spark_df = spark_df.withColumn("market_cap_usd_billion", spark_df["market_cap_usd_billion"].cast(DoubleType()))

# Show the first few rows of the PySpark DataFrame
print("\n--- Show Table Data ---")
spark_df.show(truncate=False)

# Print the schema of the DataFrame
print("--- Print Schema ---")
spark_df.printSchema()
````

##**2.2 Handle Outliers** <font color = red>[5 marks]</font> <br>


````python
# Write code for outlier analysis using IQR method
from pyspark.sql.functions import col, percentile_approx

# Calculate Q1 and Q3 for 'market_cap_usd_billion'
quantiles = spark_df.approxQuantile("market_cap_usd_billion", [0.25, 0.75], 0.05)
Q1 = quantiles[0]
Q3 = quantiles[1]

# Calculate IQR
IQR = Q3 - Q1

# Define outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print(f"Q1: {Q1}")
print(f"Q3: {Q3}")
print(f"IQR: {IQR}")
print(f"Lower Bound for Outliers: {lower_bound}")
print(f"Upper Bound for Outliers: {upper_bound}")

# Identify outliers
outliers = spark_df.filter(
    (col("market_cap_usd_billion") < lower_bound) | (col("market_cap_usd_billion") > upper_bound)
)

print("\n--- Outliers in Market Cap (US$ billion) ---")
if outliers.count() > 0:
    outliers.show(truncate=False)
else:
    print("No outliers found based on the IQR method.")
````

##**4.1. Market Capitalization Analysis** <font color = red>[3 marks]</font> <br>

Q: Perform Advanced Market Capitalization Analysis with Growth Metrics

````python
# Query: Advanced Market Cap Analysis with Growth Metrics
# Query 1: Advanced Market Capitalization Analysis with Growth Metrics

# Sort the banks by market capitalization in descending order
spark_df_sorted = spark_df.orderBy(col("market_cap_usd_billion").desc())

# Define a window specification to look at the previous row
window_spec = Window.orderBy(col("market_cap_usd_billion").desc())

# Calculate the market cap of the previous bank and the growth metric
market_cap_growth_df = spark_df_sorted.withColumn(
    "previous_market_cap",
    lag("market_cap_usd_billion", 1).over(window_spec)
).withColumn(
    "market_cap_change",
    col("market_cap_usd_billion") - col("previous_market_cap")
).withColumn(
    "market_cap_growth_percent",
    round((col("market_cap_change") / col("previous_market_cap")) * 100, 2)
)

print("--- Advanced Market Capitalization Analysis with Growth Metrics ---")
market_cap_growth_df.show(truncate=False)
````

##4.2.**Market Concentration Analysis** <font color = red>[6 marks]</font> <br>

Q: Analyze Market Concentration and Categorize Banks Based on Market Share Tiers

````python
# Market Concentration Analysis


# Calculate total market capitalization
total_market_cap = spark_df.agg(sum("market_cap_usd_billion")).collect()[0][0]

# Calculate individual and cumulative market share
market_concentration_df = spark_df.withColumn("market_share", (col("market_cap_usd_billion") / total_market_cap) * 100)

window_spec_market_share = Window.orderBy(col("market_cap_usd_billion").desc())

market_concentration_df = market_concentration_df.withColumn(
    "cumulative_market_share",
    sum("market_share").over(window_spec_market_share)
)

# Categorize banks into market share tiers
market_concentration_df = market_concentration_df.withColumn(
    "market_share_tier",
    when(col("market_share") > 15, "Tier 1 (Dominant)")
    .when(col("market_share") > 7, "Tier 2 (Major)")
    .otherwise("Tier 3 (Significant)")
)

print("--- Market Concentration and Tiers Analysis ---")
market_concentration_df.show(truncate=False)
````

##**4.4. Comparative Size Analysis** <font color = red>[6 marks]</font> <br>

Q: Conduct Comparative Size Analysis to Classify Banks by Relative Market Size

````python
# Comparative Size Analysis

# Define the market size categories based on market capitalization
comparative_size_df = spark_df.withColumn(
    "market_size_category",
    when(col("market_cap_usd_billion") > 250, "Giant")
    .when((col("market_cap_usd_billion") > 150) & (col("market_cap_usd_billion") <= 250), "Major Player")
    .when((col("market_cap_usd_billion") > 50) & (col("market_cap_usd_billion") <= 150), "Mid-Tier")
    .otherwise("Niche Player")
)

print("--- Comparative Size Analysis: Banks Classified by Market Size ---")
comparative_size_df.show(truncate=False)

# Optionally, count the number of banks in each category
print("\n--- Count of Banks by Market Size Category ---")
comparative_size_df.groupBy("market_size_category").count().orderBy("count", ascending=False).show(truncate=False)
````

##**4.5. Market Growth Analysis** <font color = red>[6 marks]</font> <br>

Q: Evaluate Market Growth and Identify Gaps Between Consecutive Banks

````python
# Growth and Gap Analysis

# Create a window specification ordered by market capitalization descending
window_spec_growth = Window.orderBy(col("market_cap_usd_billion").desc())

market_growth_df = spark_df.withColumn(
    "next_market_cap",
    lead("market_cap_usd_billion", 1).over(window_spec_growth)
).withColumn(
    "market_cap_gap_to_next",
    col("market_cap_usd_billion") - col("next_market_cap")
)

print("--- Market Growth Analysis: Gaps Between Consecutive Banks ---")
market_growth_df.show(truncate=False)
````

##**4.6. Market Dominance Analysis** <font color = red>[6 marks]</font> <br>

Q: Assess Market Dominance by Measuring Cumulative Share and Dominance Score

````python
# Market Dominance Analysis

# Calculate total market capitalization
total_market_cap_dominance = spark_df.agg(sum("market_cap_usd_billion")).collect()[0][0]

# Create a window specification ordered by market capitalization descending
window_spec_dominance = Window.orderBy(col("market_cap_usd_billion").desc())

market_dominance_df = spark_df.withColumn(
    "market_share_percent",
    (col("market_cap_usd_billion") / total_market_cap_dominance) * 100
).withColumn(
    "cumulative_market_share_percent",
    sum("market_share_percent").over(window_spec_dominance)
).withColumn(
    "dominance_score",
    (col("market_cap_usd_billion") / total_market_cap_dominance) * 1000 # Scaling for a more readable score
)

print("--- Market Dominance Analysis: Cumulative Share and Dominance Score ---")
market_dominance_df.show(truncate=False)
````

##**4.7. Segment-Wise Bank Analysis** <font color = red>[6 marks]</font> <br>

Q: Analyze Segment-Wise Bank Performance Based on Market Capitalization Ranges

````python
# Segment Performance Analysis

# Ensure the 'market_size_category' column exists. If it was not created or stored, recreate it.
# For continuity, we'll use the 'comparative_size_df' which already has this category.
# If 'comparative_size_df' is not available, you would need to recreate the categories like so:
# comparative_size_df = spark_df.withColumn(
#     "market_size_category",
#     when(col("market_cap_usd_billion") > 250, "Giant")
#     .when((col("market_cap_usd_billion") > 150) & (col("market_cap_usd_billion") <= 250), "Major Player")
#     .when((col("market_cap_usd_billion") > 50) & (col("market_cap_usd_billion") <= 150), "Mid-Tier")
#     .otherwise("Niche Player")
# )

segment_performance_df = comparative_size_df.groupBy("market_size_category").agg(
    count("bank_name").alias("number_of_banks"),
    round(sum("market_cap_usd_billion"), 2).alias("total_market_cap_billion"),
    round(avg("market_cap_usd_billion"), 2).alias("average_market_cap_billion"),
    round(min("market_cap_usd_billion"), 2).alias("min_market_cap_billion"),
    round(max("market_cap_usd_billion"), 2).alias("max_market_cap_billion")
).orderBy("total_market_cap_billion", ascending=False)

print("--- Segment-Wise Bank Performance Based on Market Capitalization Ranges ---")
segment_performance_df.show(truncate=False)
````
